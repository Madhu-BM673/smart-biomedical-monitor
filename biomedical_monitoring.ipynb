```python
import csv

heart_rates = []
temperatures = []
spo2_values = []

with open("readings.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        heart_rates.append(float(row["heart_rate"]))
        temperatures.append(float(row["temperature"]))
        spo2_values.append(float(row["spo2"]))

print("Biomedical Data Analysis")
print("----------------------")

print("Average Heart Rate:", sum(heart_rates) / len(heart_rates), "BPM")
print("Average Temperature:", sum(temperatures) / len(temperatures), "°C")
print("Average SpO2:", sum(spo2_values) / len(spo2_values), "%")

print("Highest Heart Rate:", max(heart_rates), "BPM")
print("Lowest Heart Rate:", min(heart_rates), "BPM")
```

```python
import csv
import matplotlib.pyplot as plt

# Function to read CSV data
def read_data(filename):
    heart_rates = []
    temperatures = []
    spo2_values = []

    with open(filename, "r") as file:
        reader = csv.DictReader(file)

        for row in reader:
            heart_rates.append(float(row["heart_rate"]))
            temperatures.append(float(row["temperature"]))
            spo2_values.append(float(row["spo2"]))

    return heart_rates, temperatures, spo2_values


# Function to display averages
def display_average(hr, temp, spo2):
    print("Biomedical Data Report")
    print("--------------------")
    print("Average Heart Rate:", sum(hr) / len(hr), "BPM")
    print("Average Temperature:", sum(temp) / len(temp), "°C")
    print("Average SpO2:", sum(spo2) / len(spo2), "%")


# Function to plot graphs
def plot_graphs(hr, temp, spo2):
    readings = range(1, len(hr) + 1)

    fig, axes = plt.subplots(3, 1, figsize=(9, 9))

    axes[0].plot(readings, hr, marker="o")
    axes[0].set_title("Heart Rate")
    axes[0].set_ylabel("BPM")
    axes[0].grid(True)

    axes[1].plot(readings, temp, marker="o")
    axes[1].set_title("Temperature")
    axes[1].set_ylabel("°C")
    axes[1].grid(True)

    axes[2].plot(readings, spo2, marker="o")
    axes[2].set_title("SpO2")
    axes[2].set_ylabel("%")
    axes[2].set_xlabel("Reading Number")
    axes[2].grid(True)

    plt.tight_layout()
    plt.show()


# Function to check demo alerts
def check_alerts(hr, temp, spo2):
    for i in range(len(hr)):
        print(f"\nReading {i + 1}")

        if hr[i] > 100:
            print("High Heart Rate Alert")
        elif hr[i] < 60:
            print("Low Heart Rate Alert")
        else:
            print("Heart Rate: Within demo limits")

        if temp[i] > 37.5:
            print("High Temperature Alert")
        else:
            print("Temperature: Within demo limits")

        if spo2[i] < 95:
            print("Low SpO2 Alert")
        else:
            print("SpO2: Within demo limits")


# Main program
heart_rates, temperatures, spo2_values = read_data("readings.csv")

display_average(heart_rates, temperatures, spo2_values)

plot_graphs(heart_rates, temperatures, spo2_values)

check_alerts(heart_rates, temperatures, spo2_values)
```
