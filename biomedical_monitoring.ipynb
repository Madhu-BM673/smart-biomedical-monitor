```python
import csv

heart_rates = []
temperatures = []
spo2_values = []

with open("readings.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        heart_rates.append(float(row["heart_rate"]))
        temperatures.append(float(row["temperature"]))
        spo2_values.append(float(row["spo2"]))

print("Biomedical Data Analysis")
print("----------------------")

print("Average Heart Rate:", sum(heart_rates) / len(heart_rates), "BPM")
print("Average Temperature:", sum(temperatures) / len(temperatures), "°C")
print("Average SpO2:", sum(spo2_values) / len(spo2_values), "%")

print("Highest Heart Rate:", max(heart_rates), "BPM")
print("Lowest Heart Rate:", min(heart_rates), "BPM")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(heart_rates, marker="o", label="Heart Rate")
plt.title("Heart Rate Readings")
plt.xlabel("Reading Number")
plt.ylabel("Heart Rate (BPM)")
plt.grid(True)
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

reading_numbers = range(1, len(heart_rates) + 1)

fig, axes = plt.subplots(3, 1, figsize=(10, 10))

# Heart Rate Graph
axes[0].plot(reading_numbers, heart_rates, marker="o")
axes[0].set_title("Heart Rate")
axes[0].set_ylabel("BPM")
axes[0].grid(True)

# Temperature Graph
axes[1].plot(reading_numbers, temperatures, marker="o")
axes[1].set_title("Temperature")
axes[1].set_ylabel("°C")
axes[1].grid(True)

# SpO2 Graph
axes[2].plot(reading_numbers, spo2_values, marker="o")
axes[2].set_title("SpO2")
axes[2].set_ylabel("%")
axes[2].set_xlabel("Reading Number")
axes[2].grid(True)

plt.tight_layout()
plt.show()
```

```python
print("Biomedical Alert System")
print("---------------------")

for i in range(len(heart_rates)):

    hr = heart_rates[i]
    temp = temperatures[i]
    spo2 = spo2_values[i]

    print(f"\nReading {i + 1}")

    if hr > 100:
        print("Heart Rate: High reading alert")
    elif hr < 60:
        print("Heart Rate: Low reading alert")
    else:
        print("Heart Rate: Within demo limits")

    if temp > 37.5:
        print("Temperature: High reading alert")
    else:
        print("Temperature: Within demo limits")

    if spo2 < 95:
        print("SpO2: Low reading alert")
    else:
        print("SpO2: Within demo limits")
```
